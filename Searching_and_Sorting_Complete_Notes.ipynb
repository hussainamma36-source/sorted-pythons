# Searching & Sorting Algorithms

# 1. What is Searching?

Searching means finding whether a target value exists in a collection and, usually, finding its position.

Example:

```text
arr = [10, 25, 30, 45, 50]
target = 45

Answer → index 3
```

Two fundamental approaches:

| Algorithm | Main idea | Requirement |
|---|---|---|
| Linear Search | Check elements one by one | No sorting required |
| Binary Search | Repeatedly remove half of the search space | Data/search space must be ordered |

### Interview thought process

When you see a search problem, ask:

1. Is the data sorted?
2. Is the data guaranteed to remain sorted?
3. Do I need an index, a value, or just True/False?
4. Can I eliminate half of the search space?
5. Is there a duplicate-value issue?

# 2. Linear Search

## Idea

Start from the first element and compare each element with the target.

```text
[10, 25, 30, 45, 50]
  ↑
check 10

[10, 25, 30, 45, 50]
      ↑
check 25

[10, 25, 30, 45, 50]
          ↑
check 30

[10, 25, 30, 45, 50]
              ↑
check 45 → FOUND
```

### Complexity

- Best case: **O(1)** — target is first.
- Average case: **O(n)**
- Worst case: **O(n)** — target is last or absent.
- Extra space: **O(1)**

### Important

Linear search works on **sorted or unsorted** data.

````python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i # index value
    return -1

arr = [10, 25, 30, 45, 50]

print(linear_search(arr, 45))
print(linear_search(arr, 99))
````

## Linear Search — common variations

### Find all occurrences

If duplicates are possible, do not stop at the first match when the question asks for all positions.

````python
def find_all(arr, target):
    result = []

    for i in range(len(arr)):
        if arr[i] == target:
            result.append(i)

    return result

print(find_all([2, 5, 2, 8, 2, 9], 2))
````

### Interview hack — Linear Search

If the question says:

- "first occurrence" → scan left to right and return immediately.
- "last occurrence" → scan through all elements and keep updating the answer.
- "all occurrences" → collect indices.
- "unsorted array" + simple lookup → Linear Search is usually the natural solution.

Do not use Binary Search just because it is theoretically faster if the input is not sorted.

# 3. Binary Search

## Core idea

Binary Search repeatedly cuts the search space approximately in half.

For a sorted array:

```text
[10, 20, 30, 40, 50, 60, 70]
             ↑
            mid

target = 60

60 > 40
→ Ignore the left half

[50, 60, 70]
     ↑
    mid

60 == 60 → FOUND
```

### Three variables

```python
low
high
mid
```

Usually:

```python
mid = (low + high) // 2
```

Then:

```text
target == arr[mid] → found
target > arr[mid]  → search right
target < arr[mid]  → search left
```

### Complexity

- Best: **O(1)**
- Average: **O(log n)**
- Worst: **O(log n)**
- Iterative extra space: **O(1)**

### Why O(log n)?

If there are 16 elements:

```text
16 → 8 → 4 → 2 → 1
```

Only about 4 decisions are needed.

In general, the number of steps is approximately:

```text
log₂(n)
```

````python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif target > arr[mid]:
            low = mid + 1
        else:
            high = mid - 1

    return -1

arr = [10, 20, 30, 40, 50, 60, 70]

print(binary_search(arr, 60))
print(binary_search(arr, 99))
````

# 4. Binary Search — The Most Important Template

Memorize the **logic**, not just the code.

```python
low = 0
high = len(arr) - 1

while low <= high:
    mid = (low + high) // 2

    if condition(mid):
        return answer

    elif need_right_side:
        low = mid + 1

    else:
        high = mid - 1

return not_found
```

## Common mistakes

### Mistake 1: Using `low < high` blindly

For the standard exact-search template, use:

```python
while low <= high:
```

because when `low == high`, there is still one candidate to check.

### Mistake 2: Wrong boundary update

Use:

```python
low = mid + 1
high = mid - 1
```

not:

```python
low = mid
high = mid
```

Otherwise the search can get stuck.

### Mistake 3: Forgetting the sorted requirement

Classic Binary Search assumes the search space has an order that lets you eliminate one side.

### Overflow note

In languages such as Java/C++, a safer midpoint is often:

```text
mid = low + (high - low) // 2
```

Python integers do not have the same fixed-width overflow issue, but this formula is still a good habit.

# 5. Binary Search with Duplicates

Suppose:

```text
arr = [1, 2, 2, 2, 5]
```

Target = `2`

A normal Binary Search may return **any one** of the matching indices.

But interviewers may ask:

- Find the **first occurrence**
- Find the **last occurrence**
- Find the insertion position

The trick is:

> When you find the target, do not always stop. Save the answer and continue searching in the direction that could contain a better answer.

````python
def first_occurrence(arr, target):
    low = 0
    high = len(arr) - 1
    answer = -1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            answer = mid
            high = mid - 1       # look further left
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return answer


def last_occurrence(arr, target):
    low = 0
    high = len(arr) - 1
    answer = -1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            answer = mid
            low = mid + 1        # look further right
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return answer


arr = [1, 2, 2, 2, 5]
print(first_occurrence(arr, 2))
print(last_occurrence(arr, 2))
````

# 7. Sorting

Sorting means arranging data in an order.

### Ascending

```text
[50, 10, 40, 20, 30]
       ↓
[10, 20, 30, 40, 50]
```

### Descending

```text
[50, 10, 40, 20, 30]
       ↓
[50, 40, 30, 20, 10]
```

Why do we sort?

- Makes searching easier.
- Enables Binary Search.
- Helps detect duplicates.
- Helps solve interval and two-pointer problems.
- Makes greedy strategies possible in many problems.
- Makes data easier to process systematically.

# 8. Bubble Sort

## Idea

Compare adjacent elements and swap them if they are in the wrong order.

Example:

```text
[5, 3, 4, 1]

5 > 3 → swap
[3, 5, 4, 1]

5 > 4 → swap
[3, 4, 5, 1]

5 > 1 → swap
[3, 4, 1, 5]
```

The largest unsorted element moves toward the end — like a bubble rising.

### Complexity

Basic version:

- Best: O(n²)
- Average: O(n²)
- Worst: O(n²)
- Space: O(1)

Optimized version with a swap flag:

- Best: O(n)
- Average: O(n²)
- Worst: O(n²)

````python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr


arr = [5, 3, 4, 1, 2]
print(bubble_sort(arr))
````

### Bubble Sort interview point

Bubble Sort is mainly useful for learning:

- nested loops
- adjacent comparison
- swapping
- optimization using a flag

In real Python coding interviews, you usually should **not choose Bubble Sort** for a large input when an O(n log n) solution is appropriate.

# 9. Selection Sort

## Idea

Find the smallest element from the unsorted portion and place it at the beginning.

```text
[5, 3, 4, 1, 2]
 ↑

minimum = 1

[1, 3, 4, 5, 2]
```

Then repeat for the remaining portion.

### Complexity

- Best: O(n²)
- Average: O(n²)
- Worst: O(n²)
- Space: O(1)

Unlike optimized Bubble Sort, Selection Sort still performs O(n²) comparisons even when the array is already sorted.

````python
def selection_sort(arr):
    n = len(arr)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

    return arr


arr = [5, 3, 4, 1, 2]
print(selection_sort(arr))
````

# 10. Insertion Sort

Think of arranging playing cards in your hand.

The left side is kept sorted. Take the next element and insert it into the correct position.

```text
[5, 3, 4, 1]

Start:
[5]

Insert 3:
[3, 5]

Insert 4:
[3, 4, 5]

Insert 1:
[1, 3, 4, 5]
```

### Complexity

- Best: O(n) — already/almost sorted
- Average: O(n²)
- Worst: O(n²)
- Space: O(1)

### When is it useful?

Insertion Sort is good for:

- small arrays
- nearly sorted data
- understanding incremental insertion

````python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


arr = [5, 3, 4, 1, 2]
print(insertion_sort(arr))
````

# 11. Merge Sort

Merge Sort uses **Divide and Conquer**.

### Three stages

1. Divide the array.
2. Recursively sort each half.
3. Merge the sorted halves.

```text
              [8, 4, 2, 6, 1, 5]
                    /       \
             [8,4,2]       [6,1,5]
              /   \         /   \
            ...   ...      ...   ...
                    ↓
             sorted halves
                    ↓
             [1,2,4,5,6,8]
```

### Complexity

- Best: O(n log n)
- Average: O(n log n)
- Worst: O(n log n)
- Extra space: O(n) for the typical array implementation

### Interview concept

Merge Sort gives predictable O(n log n) time, but it generally needs additional memory for arrays.

````python
def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


arr = [8, 4, 2, 6, 1, 5]
print(merge_sort(arr))
````

# 12. Quick Sort

Quick Sort also follows **Divide and Conquer**.

It chooses a **pivot**, partitions the data around the pivot, then recursively sorts the partitions.

Example:

```text
[7, 2, 9, 4, 1, 5]

pivot = 5

smaller: [2, 4, 1]
pivot:   [5]
larger:  [7, 9]
```

Then sort the smaller and larger sections.

### Complexity

- Best: O(n log n)
- Average: O(n log n)
- Worst: O(n²)
- Extra space: depends on implementation/recursion; in-place versions can use O(log n) average recursion stack and O(n) worst-case stack.

### Why can worst case be O(n²)?

If the pivot repeatedly produces a very unbalanced partition:

```text
n elements
→ n-1 and 0
→ n-2 and 0
→ n-3 and 0
...
```

Pivot selection matters.

````python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)


arr = [7, 2, 9, 4, 1, 5, 5]
print(quick_sort(arr))
````

# 13. Python Sorting — Very Important

In real Python programs, prefer the built-in sorting tools unless the interviewer specifically asks you to implement a sorting algorithm.

## `sorted()`

Returns a **new list**.

```python
arr = [5, 2, 4, 1]

new_arr = sorted(arr)

print(arr)
print(new_arr)
```

## `.sort()`

Sorts the list **in place** and returns `None`.

```python
arr = [5, 2, 4, 1]
arr.sort()

print(arr)
```

### Complexity

Python's list sorting uses **Timsort**, which has:

- Worst-case: O(n log n)
- Best-case behavior can approach O(n) on suitable partially ordered data.
- It is stable.

````python
arr = [5, 2, 4, 1]

a = sorted(arr)
print("original:", arr)
print("sorted:", a)

arr.sort()
print("in-place:", arr)
````

# 15. Sorting Algorithms — Quick Comparison

| Algorithm | Best | Average | Worst | Extra Space | Key idea |
|---|---:|---:|---:|---:|---|
| Bubble | O(n)* | O(n²) | O(n²) | O(1) | Adjacent swaps |
| Selection | O(n²) | O(n²) | O(n²) | O(1) | Select minimum |
| Insertion | O(n) | O(n²) | O(n²) | O(1) | Insert into sorted part |
| Merge | O(n log n) | O(n log n) | O(n log n) | O(n) typical | Divide + merge |
| Quick | O(n log n) | O(n log n) | O(n²) | Depends | Pivot + partition |
| Python `sort()` | ~O(n) on favorable runs | O(n log n) | O(n log n) | Implementation-dependent | Timsort |

`*` with the optimized swap-flag version.

### Interview shortcut

For large input:

```text
O(n²)  → usually suspicious
O(n log n) → commonly acceptable
O(log n) → very efficient
O(1) → constant
```

Always consider the actual constraints before deciding.

# 16. Searching + Sorting Relationship

A very common pattern is:

```text
Unsorted data
     ↓
Sort O(n log n)
     ↓
Binary Search O(log n) per query
```

If there are many search queries, sorting once can be worthwhile.

Example:

```text
Array size = n
Queries = q

Sort once: O(n log n)
Each binary search: O(log n)
Total: O(n log n + q log n)
```

Compare that with Linear Search for every query:

```text
O(nq)
```

For large `q`, the difference can be huge.

> **Interview hack:** If you see "many queries on the same array", think about preprocessing — sorting, hashing, prefix sums, etc.

# 17. Searching Patterns Beyond Basic Binary Search

Binary Search is not only for "find target in sorted array".

Look for phrases such as:

- minimum possible
- maximum possible
- first valid
- last valid
- smallest value satisfying condition
- largest value satisfying condition
- minimize the maximum
- maximize the minimum

These often indicate **Binary Search on Answer**.

### Mental model

You are searching a range of possible answers:

```text
possible answers
[low ---------------------- high]

        ↓ binary search

Is this answer possible?

YES → maybe search smaller
NO  → search larger
```

The key requirement is a **monotonic predicate**:

```text
False False False True True True
```

or

```text
True True True False False False
```

Once the condition changes, it should not change back and forth.

# 19. Interview Hacks — Sorting

### Hack 1 — Need unique values?

Consider:

```python
set(arr)
```

But remember: a set does not preserve the ordering needed for Binary Search.

### Hack 2 — Need frequency?

Consider a dictionary / `Counter` before sorting.

### Hack 3 — Need two-pointer technique?

Sorting is often the first step.

Example:

```text
[2, 7, 11, 15]
target = 9

left → 2
right → 15

sum > target → move right
sum < target → move left
```

### Hack 4 — Intervals?

Sort by starting time.

Typical pattern:

```python
intervals.sort(key=lambda x: x[0])
```

### Hack 5 — Need top K?

Do not automatically sort the entire array. Consider:

- heap
- Quickselect
- frequency counting

depending on constraints.

### Hack 6 — Need to compare two arrays?

Sorting both can turn many comparison problems into simple two-pointer scans.

# 21. Complexity Cheat Sheet

| Complexity | Meaning | Typical example |
|---|---|---|
| O(1) | Constant | Array index access |
| O(log n) | Halves search space | Binary Search |
| O(n) | One pass | Linear Search |
| O(n log n) | Efficient sorting | Merge Sort / Python sort |
| O(n²) | Nested comparisons | Bubble / Selection / Insertion worst case |
| O(2ⁿ) | Exponential | Many brute-force recursion problems |
| O(n!) | Factorial | Permutation brute force |

### Golden interview rule

Do not just say:

> "This is fast."

Say:

> "Time complexity is O(n log n) and extra space is O(n)."

# 22. Practice: Identify the Algorithm

### Question 1
Array is unsorted and you need one element.

**Likely:** Linear Search.

### Question 2
Array is sorted and you need one element.

**Likely:** Binary Search.

### Question 3
You need the first position where `x` can be inserted.

**Likely:** Lower-bound style Binary Search / `bisect_left`.

### Question 4
You need to merge overlapping intervals.

**Likely:** Sort intervals first, then scan.

### Question 5
You need to find whether two numbers sum to a target.

Possible approaches:

- Hash Set/Dictionary: O(n) average.
- Sort + two pointers: O(n log n).

Choose based on whether modifying/order matters and what the problem asks.

You need the kth largest element.

Do not automatically sort. Consider:

- heap
- Quickselect
- sorting, if constraints allow

Do not memorize 7 algorithms as isolated programs.

1. **Linear Search** → scan.
2. **Binary Search** → halve the search space.
3. **Bubble Sort** → adjacent swaps.
4. **Selection Sort** → choose minimum.
5. **Insertion Sort** → insert into sorted portion.
6. **Merge Sort** → divide and merge.
7. **Quick Sort** → pivot and partition.
8. **Python sort** → use it in real applications unless implementation is requested.
9. **Sort + Two Pointers** → extremely common interview pattern.
10. **Binary Search on Answer** → extremely important advanced pattern.

> **Interview mindset:** Don't ask only "Which algorithm do I know?" Ask **"What property of this problem can I exploit?"**
